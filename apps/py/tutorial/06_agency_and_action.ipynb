# Agency & Action: Who Does Things in a Text?

Motifs tell us what a text talks *about*. But texts also have **agents**—entities that speak, act, command, and suffer.

This notebook introduces simple ways to track:
- **Who speaks** (dialogue attribution)
- **Who acts** (subjects of verbs)
- **Who commands** (imperatives)
- **Who is acted upon** (objects, patients)

---

## The Question

In a creation narrative, who has agency?

- Does God act alone, or with others?
- Do creatures have agency, or are they passive?
- Who speaks, and who is spoken to?

## 1. Setup

```python
from collections import Counter
import re

# Genesis 1 with punctuation preserved for speech detection
genesis_1 = """In the beginning God created the heaven and the earth.
And the earth was without form, and void; and darkness was upon the face of the deep.
And the Spirit of God moved upon the face of the waters.
And God said, Let there be light: and there was light.
And God saw the light, that it was good: and God divided the light from the darkness.
And God called the light Day, and the darkness he called Night.
And the evening and the morning were the first day.
And God said, Let there be a firmament in the midst of the waters, and let it divide the waters from the waters.
And God made the firmament, and divided the waters which were under the firmament from the waters which were above the firmament: and it was so.
And God called the firmament Heaven. And the evening and the morning were the second day.
And God said, Let the waters under the heaven be gathered together unto one place, and let the dry land appear: and it was so.
And God called the dry land Earth; and the gathering together of the waters called he Seas: and God saw that it was good.
And God said, Let the earth bring forth grass, the herb yielding seed, and the fruit tree yielding fruit after his kind, whose seed is in itself, upon the earth: and it was so.
And the earth brought forth grass, and herb yielding seed after his kind, and the tree yielding fruit, whose seed was in itself, after his kind: and God saw that it was good.
And the evening and the morning were the third day.
And God said, Let there be lights in the firmament of the heaven to divide the day from the night; and let them be for signs, and for seasons, and for days, and years.
And let them be for lights in the firmament of the heaven to give light upon the earth: and it was so.
And God made two great lights; the greater light to rule the day, and the lesser light to rule the night: he made the stars also.
And God set them in the firmament of the heaven to give light upon the earth.
And to rule over the day and over the night, and to divide the light from the darkness: and God saw that it was good.
And the evening and the morning were the fourth day.
And God said, Let the waters bring forth abundantly the moving creature that hath life, and fowl that may fly above the earth in the open firmament of heaven.
And God created great whales, and every living creature that moveth, which the waters brought forth abundantly, after their kind, and every winged fowl after his kind: and God saw that it was good.
And God blessed them, saying, Be fruitful, and multiply, and fill the waters in the seas, and let fowl multiply in the earth.
And the evening and the morning were the fifth day.
And God said, Let the earth bring forth the living creature after his kind, cattle, and creeping thing, and beast of the earth after his kind: and it was so.
And God made the beast of the earth after his kind, and cattle after their kind, and every thing that creepeth upon the earth after his kind: and God saw that it was good.
And God said, Let us make man in our image, after our likeness: and let them have dominion over the fish of the sea, and over the fowl of the air, and over the cattle, and over all the earth, and over every creeping thing that creepeth upon the earth.
So God created man in his own image, in the image of God created he him; male and female created he them.
And God blessed them, and God said unto them, Be fruitful, and multiply, and replenish the earth, and subdue it: and have dominion over the fish of the sea, and over the fowl of the air, and over every living thing that moveth upon the earth."""

print(f"Loaded {len(genesis_1)} characters")
```

## 2. Finding Speech Acts

In many texts, speech is marked by patterns like "X said" or "X spoke".

Let's find who speaks.

```python
# Speech verbs to look for
SPEECH_VERBS = ["said", "says", "spoke", "speak", "cried", "called", "answered", "replied"]

def find_speakers(text):
    """Find patterns like 'X said' or 'X spoke'."""
    speakers = []
    
    # Pattern: word + speech verb
    for verb in SPEECH_VERBS:
        # Look for "[Word] [verb]" pattern
        pattern = r'(\b[A-Z][a-z]+\b)\s+' + verb
        matches = re.findall(pattern, text)
        for match in matches:
            speakers.append((match, verb))
    
    return speakers

speakers = find_speakers(genesis_1)
print("Speech acts found:")
for speaker, verb in speakers:
    print(f"  {speaker} {verb}")

print(f"\nTotal speech acts: {len(speakers)}")
print(f"Unique speakers: {set(s for s, v in speakers)}")
```

```python
# Count speakers
speaker_counts = Counter(s for s, v in speakers)
print("Speaker frequency:")
for speaker, count in speaker_counts.most_common():
    print(f"  {speaker}: {count} speech acts")
```

### What do you notice?

In Genesis 1, **God** is the only speaker. The text is a monologue of divine speech acts.

This is a structural feature: creation happens through speech ("Let there be...").

---

## 3. Finding Action Verbs

Beyond speech, who *does* things? Let's look for subjects of action verbs.

```python
# Action verbs relevant to creation narratives
ACTION_VERBS = [
    "created", "made", "divided", "called", "set", "blessed",
    "moved", "brought", "gathered", "ruled", "gave"
]

def find_actors(text, verbs):
    """Find patterns like 'X [verb]'."""
    actors = []
    
    for verb in verbs:
        # Look for "[Word] [verb]" pattern
        pattern = r'(\b[A-Z][a-z]+\b)\s+' + verb
        matches = re.findall(pattern, text)
        for match in matches:
            actors.append((match, verb))
    
    return actors

actors = find_actors(genesis_1, ACTION_VERBS)
print("Action patterns found:")
for actor, verb in actors:
    print(f"  {actor} {verb}")
```

```python
# Count actors
actor_counts = Counter(a for a, v in actors)
print("\nActor frequency:")
for actor, count in actor_counts.most_common():
    print(f"  {actor}: {count} actions")

# What verbs does each actor use?
print("\nVerbs by actor:")
for actor in actor_counts.keys():
    verbs = [v for a, v in actors if a == actor]
    print(f"  {actor}: {Counter(verbs)}")
```

### Interpretation

God dominates the action verbs: created, made, divided, called, set, blessed.

But notice: the **earth** also acts ("brought forth"). This is interesting—the earth has limited agency in the creation process.

## 4. Commands and Imperatives

Commands are a special kind of speech act. They direct action.

In Genesis 1, God's speech is largely imperative: "Let there be..."

```python
# Find imperative patterns
def find_commands(text):
    """Find imperative patterns like 'Let X' or 'Be X'."""
    commands = []
    
    # "Let there be X" pattern
    let_patterns = re.findall(r'Let there be ([a-z]+)', text)
    for match in let_patterns:
        commands.append(("Let there be", match))
    
    # "Let X [verb]" pattern
    let_verb = re.findall(r'Let the ([a-z]+) ([a-z]+)', text)
    for noun, verb in let_verb:
        commands.append((f"Let the {noun}", verb))
    
    # "Be [adjective]" pattern
    be_patterns = re.findall(r'Be ([a-z]+)', text)
    for match in be_patterns:
        commands.append(("Be", match))
    
    return commands

commands = find_commands(genesis_1)
print("Commands found:")
for cmd_type, target in commands:
    print(f"  {cmd_type} → {target}")
```

```python
# What is commanded into existence?
commanded_things = [target for cmd, target in commands if cmd == "Let there be"]
print("Things commanded into existence:")
for thing in commanded_things:
    print(f"  {thing}")

# What is commanded to act?
commanded_actors = [(noun, verb) for cmd, verb in commands 
                    if cmd.startswith("Let the") 
                    for noun in [cmd.replace("Let the ", "")]]
print("\nThings commanded to act:")
for noun, verb in commanded_actors:
    print(f"  {noun} → {verb}")
```

### Reflection

The command structure reveals a theology:
- God commands, things obey
- Some things are commanded into existence ("Let there be light")
- Some things are commanded to act ("Let the earth bring forth")

The earth, waters, and creatures have **delegated agency**—they act, but only when commanded.

## 5. Passive vs Active Voice

Who acts vs who is acted upon?

Passive constructions ("was created", "were divided") hide agency. Active constructions ("God created", "God divided") make it visible.

```python
# Simple passive detection: "was/were [verb]ed"
def find_passives(text):
    """Find passive voice patterns."""
    passives = []
    
    # "was [verb]ed" pattern
    was_patterns = re.findall(r'was (\w+ed)', text.lower())
    passives.extend([("was", v) for v in was_patterns])
    
    # "were [verb]ed" pattern  
    were_patterns = re.findall(r'were (\w+ed)', text.lower())
    passives.extend([("were", v) for v in were_patterns])
    
    return passives

passives = find_passives(genesis_1)
print(f"Passive constructions found: {len(passives)}")
for aux, verb in passives[:10]:
    print(f"  {aux} {verb}")
```

```python
# Compare: how many active vs passive constructions?
active_count = len(actors)
passive_count = len(passives)

print(f"Active constructions (X did Y): {active_count}")
print(f"Passive constructions (Y was done): {passive_count}")
print(f"\nRatio: {active_count / max(passive_count, 1):.1f}x more active than passive")
```

### Interpretation

Genesis 1 is heavily **active voice**. God is named as the agent repeatedly.

Compare this to scientific writing, which often uses passive voice to hide human agency ("the experiment was conducted" vs "we conducted the experiment").

## 6. Agency Profile

Let's combine these measures into an "agency profile" for the text.

```python
def agency_profile(text):
    """Build an agency profile for a text."""
    speakers = find_speakers(text)
    actors = find_actors(text, ACTION_VERBS)
    commands = find_commands(text)
    passives = find_passives(text)
    
    profile = {
        "total_speech_acts": len(speakers),
        "unique_speakers": list(set(s for s, v in speakers)),
        "speaker_counts": dict(Counter(s for s, v in speakers)),
        
        "total_actions": len(actors),
        "unique_actors": list(set(a for a, v in actors)),
        "actor_counts": dict(Counter(a for a, v in actors)),
        
        "total_commands": len(commands),
        "command_types": dict(Counter(c for c, t in commands)),
        
        "passive_constructions": len(passives),
        "active_passive_ratio": len(actors) / max(len(passives), 1)
    }
    
    return profile

genesis_profile = agency_profile(genesis_1)

print("Agency Profile: Genesis 1")
print("=" * 40)
for key, value in genesis_profile.items():
    print(f"{key}: {value}")
```

## 7. Comparing Agency Across Texts

Let's compare Genesis to another creation text.

```python
# Prometheus myth (Hesiod, paraphrased)
prometheus = """Zeus said to Prometheus, You have stolen fire from the gods.
Prometheus answered, I gave fire to mortals, for they suffered in darkness.
Zeus commanded the gods, Bind Prometheus to the rock.
Hephaestus made the chains and bound Prometheus.
The eagle came daily and ate the liver of Prometheus.
Prometheus endured the torment, for he knew what was to come.
Zeus said, You will suffer until one frees you.
Heracles came and killed the eagle.
Heracles freed Prometheus from his chains.
Zeus allowed this, for Heracles was his son.
Prometheus said to Heracles, I thank you, son of Zeus.
The mortals kept the fire and honored Prometheus."""

prometheus_profile = agency_profile(prometheus)

print("Agency Profile: Prometheus Myth")
print("=" * 40)
for key, value in prometheus_profile.items():
    print(f"{key}: {value}")
```

```python
# Side-by-side comparison
print("\nComparison: Genesis vs Prometheus")
print("=" * 50)
print(f"{'Measure':30} {'Genesis':>10} {'Prometheus':>10}")
print("-" * 50)

for key in ["total_speech_acts", "total_actions", "total_commands", "passive_constructions"]:
    g = genesis_profile[key]
    p = prometheus_profile[key]
    print(f"{key:30} {g:>10} {p:>10}")

print(f"\n{'Unique speakers':30} {len(genesis_profile['unique_speakers']):>10} {len(prometheus_profile['unique_speakers']):>10}")
print(f"{'Unique actors':30} {len(genesis_profile['unique_actors']):>10} {len(prometheus_profile['unique_actors']):>10}")
```

### Interpretation

The Prometheus myth has:
- **Multiple speakers** (Zeus, Prometheus, Heracles)
- **Multiple actors** (gods, Prometheus, eagle, Heracles, mortals)
- **Dialogue and conflict**

Genesis 1 has:
- **One speaker** (God)
- **One primary actor** (God)
- **Monologue and command**

These are fundamentally different models of divine agency and creation.

## 8. Limitations

This approach is **approximate**. It misses:

- **Pronouns**: "He said" doesn't tell us who "he" is
- **Complex syntax**: "The man whom God created" has nested agency
- **Implicit agents**: "The light appeared" hides who made it appear
- **Metaphorical agency**: "The earth brought forth" might be literal or figurative

These are real limitations. But even approximate measures can reveal patterns.
