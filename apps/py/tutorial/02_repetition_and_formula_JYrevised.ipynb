# Repetition & Formula: What a Text Says Over and Over

Literature repeats itself. Words, phrases, images, syntactic patterns—repetition is how texts create rhythm, emphasis, and meaning.

This notebook introduces **word frequency** as a way to see what a text insists on.

## 1. Setup: Loading Texts

We'll work with two biblical passages to compare. Both are from the King James Version and both describe creation, but they approach it very differently.

- **Genesis 1**: The creation narrative—structured, formulaic, repetitive
- **Psalm 104**: A creation hymn—poetic, varied, celebratory

By comparing them, we'll see how different frequency measures reveal different aspects of meaning.

```python
from collections import Counter
import re
import math

# Sample text 1: Genesis 1 (KJV)
genesis_1 = """
In the beginning God created the heaven and the earth.
And the earth was without form, and void; and darkness was upon the face of the deep.
And the Spirit of God moved upon the face of the waters.
And God said, Let there be light: and there was light.
And God saw the light, that it was good: and God divided the light from the darkness.
And God called the light Day, and the darkness he called Night.
And the evening and the morning were the first day.
And God said, Let there be a firmament in the midst of the waters,
and let it divide the waters from the waters.
And God made the firmament, and divided the waters which were under the firmament
from the waters which were above the firmament: and it was so.
And God called the firmament Heaven. And the evening and the morning were the second day.
And God said, Let the waters under the heaven be gathered together unto one place,
and let the dry land appear: and it was so.
And God called the dry land Earth; and the gathering together of the waters called he Seas:
and God saw that it was good.
And God said, Let the earth bring forth grass, the herb yielding seed,
and the fruit tree yielding fruit after his kind, whose seed is in itself, upon the earth: and it was so.
And the earth brought forth grass, and herb yielding seed after his kind,
and the tree yielding fruit, whose seed was in itself, after his kind: and God saw that it was good.
And the evening and the morning were the third day.
"""

# Sample text 2: Psalm 104 (KJV) 
psalm_104 = """
Bless the Lord, O my soul. O Lord my God, thou art very great; thou art clothed with honour and majesty.
Who coverest thyself with light as with a garment: who stretchest out the heavens like a curtain:
Who layeth the beams of his chambers in the waters: who maketh the clouds his chariot: who walketh upon the wings of the wind:
Who maketh his angels spirits; his ministers a flaming fire:
Who laid the foundations of the earth, that it should not be removed for ever.
Thou coveredst it with the deep as with a garment: the waters stood above the mountains.
At thy rebuke they fled; at the voice of thy thunder they hasted away.
They go up by the mountains; they go down by the valleys unto the place which thou hast founded for them.
Thou hast set a bound that they may not pass over; that they turn not again to cover the earth.
He sendeth the springs into the valleys, which run among the hills.
They give drink to every beast of the field: the wild asses quench their thirst.
By them shall the fowls of the heaven have their habitation, which sing among the branches.
He watereth the hills from his chambers: the earth is satisfied with the fruit of thy works.
He causeth the grass to grow for the cattle, and herb for the service of man: that he may bring forth food out of the earth;
And wine to make glad the heart of man, and oil to make his face to shine, and bread which strengtheneth man's heart.
"""

print(f"Genesis 1: {len(genesis_1)} characters")
print(f"Psalm 104: {len(psalm_104)} characters")
```

## 2. Counting Words

The simplest distant reading move: count how often each word appears.

```python
def get_words(text):
    """Extract words from text, lowercased, punctuation removed."""
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    words = text.split()
    return words

genesis_words = get_words(genesis_1)
psalm_words = get_words(psalm_104)

print(f"Genesis words: {len(genesis_words)}")
print(f"Psalm 104 words: {len(psalm_words)}")
```

```python
# Count word frequencies
genesis_counts = Counter(genesis_words)

print("Genesis - Top 20 most frequent words:")
for word, count in genesis_counts.most_common(20):
    print(f"  {word}: {count}")
```

### What do you notice?

The most frequent words are probably **function words**: "the", "and", "was", "of"...

These words are grammatically necessary but don't tell us much about *this* text specifically.

They're sometimes called **stopwords**.

---

## 3. Filtering Stopwords

Let's filter out common English words to reveal the **content words**—the vocabulary that carries meaning.

```python
# A simple stopword list (you can modify this!)
STOPWORDS = {
    'the', 'and', 'a', 'an', 'of', 'to', 'in', 'is', 'it', 'that', 'was', 'were',
    'be', 'been', 'being', 'have', 'has', 'had', 'do', 'does', 'did',
    'for', 'on', 'with', 'as', 'at', 'by', 'from', 'or', 'but', 'not',
    'this', 'which', 'there', 'their', 'they', 'he', 'she', 'his', 'her',
    'i', 'you', 'we', 'my', 'your', 'our', 'its', 'who', 'what', 'when',
    'where', 'why', 'how', 'all', 'each', 'every', 'both', 'few', 'more',
    'most', 'other', 'some', 'such', 'no', 'nor', 'only', 'own', 'same',
    'so', 'than', 'too', 'very', 'just', 'can', 'will', 'should', 'now',
    'unto', 'upon', 'let', 'whose', 'after', 'itself'
}

def filter_stopwords(words):
    return [w for w in words if w not in STOPWORDS]

genesis_content = filter_stopwords(genesis_words)
psalm_content = filter_stopwords(psalm_words)

print(f"Genesis - content words: {len(genesis_content)} (was {len(genesis_words)})")
print(f"Psalm 104 - content words: {len(psalm_content)} (was {len(psalm_words)})")
```

```python
# Count again without stopwords
genesis_content_counts = Counter(genesis_content)
psalm_content_counts = Counter(psalm_content)

print("Genesis - Top 15 content words:")
for word, count in genesis_content_counts.most_common(15):
    print(f"  {word}: {count}")

print("\nPsalm 104 - Top 15 content words:")
for word, count in psalm_content_counts.most_common(15):
    print(f"  {word}: {count}")
```

### Now what do you see?

These words tell us what each text is *about*:

**Genesis 1**: god, earth, waters, light, day, called, said...

**Psalm 104**: earth, waters, lord, man, mountains, hills...

They share creation vocabulary (earth, waters) because both are about creation. But there's still a problem:

**Absolute counts depend on text length.** A longer text will naturally have higher counts. What we really want to know is: what's *distinctive* about each text? What makes Genesis different from Psalm 104?

---

## 4. Finding Distinctive Words: TF-IDF

Both texts use "earth" and "waters"—they're creation texts. But when does a word become **signature vocabulary**?

**TF-IDF** (Term Frequency-Inverse Document Frequency) answers this question. It identifies words that are:
- Frequent in one text (high TF)
- Rare across all texts (high IDF)

Formula: `TF-IDF = (word frequency in document) × log(total documents / documents containing word)`

The intuition: a word that appears 10 times in Genesis but also 10 times in every other text isn't distinctive. But a word that appears 10 times in Genesis and nowhere else is *signature vocabulary*.

```python
def calculate_tfidf(documents_as_word_lists):
    """
    Calculate TF-IDF for each word in each document.
    
    TF (Term Frequency) = how often a word appears in a document
    IDF (Inverse Document Frequency) = how rare the word is across all documents
    TF-IDF = TF × IDF
    
    Returns: list of dictionaries, one per document, mapping word -> TF-IDF score
    """
    num_docs = len(documents_as_word_lists)
    
    # Calculate document frequency for each word
    # (how many documents contain this word?)
    doc_freq = Counter()
    for words in documents_as_word_lists:
        unique_words = set(words)
        for word in unique_words:
            doc_freq[word] += 1
    
    # Calculate TF-IDF for each document
    tfidf_scores = []
    for words in documents_as_word_lists:
        tf = Counter(words)  # term frequency in this doc
        tfidf = {}
        
        for word, count in tf.items():
            # TF: just the raw count (we could normalize, but let's keep it simple)
            term_freq = count
            
            # IDF: log(total docs / docs containing word)
            inverse_doc_freq = math.log(num_docs / doc_freq[word])
            
            # TF-IDF is the product
            tfidf[word] = term_freq * inverse_doc_freq
        
        tfidf_scores.append(tfidf)
    
    return tfidf_scores

# Calculate TF-IDF for both texts
documents = [genesis_content, psalm_content]
tfidf_results = calculate_tfidf(documents)

genesis_tfidf = sorted(tfidf_results[0].items(), key=lambda x: x[1], reverse=True)
psalm_tfidf = sorted(tfidf_results[1].items(), key=lambda x: x[1], reverse=True)

print("Genesis - Top 15 words by TF-IDF (distinctiveness):")
for word, score in genesis_tfidf[:15]:
    print(f"  {word}: {score:.2f}")

print("\nPsalm 104 - Top 15 words by TF-IDF (distinctiveness):")
for word, score in psalm_tfidf[:15]:
    print(f"  {word}: {score:.2f}")
```

### What This Reveals

Look at what TF-IDF highlights as distinctive:

**Genesis 1**: called, said, day, darkness, evening, morning...
- Vocabulary of **divine command and separation**
- Emphasis on **naming** ("called") and **speaking** ("said")
- Structured time markers ("evening", "morning")

**Psalm 104**: thou, thy, lord, mountains, valleys...
- Vocabulary of **divine provision and relationship**
- Second-person address ("thou", "thy")—it's a prayer
- Focus on **landscape** and **human benefit**

Both texts are about creation, but they reveal different theological emphases:
- **Genesis**: God creates through command and division
- **Psalm 104**: God sustains through provision and relationship

This is what counting reveals that close reading alone might miss—the statistical signature of different approaches to the same topic.

## 5. Additional Lenses: Beyond Individual Words

So far we've counted individual words. But texts also create meaning through:
- **Repeated phrases** (structural formulas)
- **Unique words** (lexical richness)

These measures reveal different aspects of how a text works.

### A. Repeated Phrases: Finding Formulas

Instead of counting individual words, we can count **n-grams**—sequences of n words.

Repeated phrases reveal:
- Formulaic structure
- Oral/liturgical patterns
- Textual scaffolding

Let's find the repeated phrases in Genesis 1.

```python
def get_ngrams(words, n):
    """Get all n-word sequences from a list of words."""
    ngrams = []
    for i in range(len(words) - n + 1):
        ngram = ' '.join(words[i:i+n])
        ngrams.append(ngram)
    return ngrams
```

```python
# Get bigrams (2-word phrases) from Genesis
bigrams = get_ngrams(genesis_words, 2)
bigram_counts = Counter(bigrams)

print("Genesis - Most repeated 2-word phrases:")
for phrase, count in bigram_counts.most_common(15):
    if count > 1:  # Only show repeated phrases
        print(f"  '{phrase}': {count}")
```

```python
# Get trigrams (3-word phrases)
trigrams = get_ngrams(genesis_words, 3)
trigram_counts = Counter(trigrams)

print("Genesis - Most repeated 3-word phrases:")
for phrase, count in trigram_counts.most_common(15):
    if count > 1:
        print(f"  '{phrase}': {count}")
```

### What formulas do you find?

In Genesis 1, you might see:
- "and god said"
- "it was so"
- "and god saw"
- "the evening and the morning"
- "after his kind"

These are the **structural formulas** of the text—the repeated scaffolding that organizes the narrative.

Genesis 1 is highly formulaic: each day follows a pattern (command → execution → evaluation → time marker).

Oral traditions rely heavily on formula. So do legal codes, liturgies, and certain modernist experiments (think of Gertrude Stein's repetitions).

### B. Lexical Diversity: Hapax Legomena

The opposite of repetition: words that appear **exactly once**.

In biblical studies, these are called *hapax legomena* (Greek: "said once").

The **ratio of unique words to total words** measures lexical diversity:
- High ratio = lexically diverse, experimental, varied
- Low ratio = formulaic, repetitive, focused

Neither is better—they're just different ways texts create meaning.

```python
# Find words that appear only once
genesis_hapax = [word for word, count in genesis_content_counts.items() if count == 1]
psalm_hapax = [word for word, count in psalm_content_counts.items() if count == 1]

genesis_diversity = len(set(genesis_content)) / len(genesis_content)
psalm_diversity = len(set(psalm_content)) / len(psalm_content)

print(f"Genesis:")
print(f"  Unique words: {len(set(genesis_content))}")
print(f"  Total content words: {len(genesis_content)}")
print(f"  Hapax (appearing once): {len(genesis_hapax)}")
print(f"  Lexical diversity: {genesis_diversity:.2%}\n")

print(f"Psalm 104:")
print(f"  Unique words: {len(set(psalm_content))}")
print(f"  Total content words: {len(psalm_content)}")
print(f"  Hapax (appearing once): {len(psalm_hapax)}")
print(f"  Lexical diversity: {psalm_diversity:.2%}")

print("\nGenesis hapax examples:")
print(f"  {sorted(genesis_hapax)[:15]}")
print("\nPsalm hapax examples:")
print(f"  {sorted(psalm_hapax)[:15]}")
```

### Reflection

Hapax words are:
- **Unique** to a specific moment in the text
- Often **specific** (proper nouns, technical terms)
- Sometimes **errors** or **variants**

Lexical diversity is a **relative measure**—it doesn't depend on absolute text length, making it useful for comparison.

## Summary: What Different Frequency Measures Reveal

| Method | What It Shows | Best For | Limitation |
|--------|----------|----------|------------|
| **Raw Frequency** | What words repeat most | Finding obsessions, formulas | Depends on text length |
| **Relative Frequency** | What percentage each word represents | Comparing texts of different lengths | Doesn't show what's distinctive |
| **TF-IDF** | What makes this text different | Finding signature vocabulary | Requires comparison corpus |
| **Hapax** | What appears only once | Measuring lexical richness, specificity | Misses important repeated words |
| **Repeated Phrases** | What formulas or structures recur | Finding textual scaffolding | Doesn't capture meaning |

None of these measures is "correct." Each offers a different lens for distant reading—revealing patterns invisible to close reading alone.
