# Motifs as Buckets: Grouping Words Is Interpretation

In the last notebook, we counted individual words. But words don't mean alone—they cluster.

**"light"**, **"sun"**, **"bright"**, **"shine"**, **"glow"** might all belong to the same *motif*.

This notebook introduces **motif buckets**: lists of related words that we group together as an interpretive act.

---

## The Key Idea

A motif bucket is a **theory**. It says: "These words belong together because they share something."

Different scholars might draw the buckets differently. That's not a bug—it's the point.

**Grouping is interpretation.**

---

## 1. Setup

```python
from collections import Counter
import re

def get_words(text):
    """Extract words from text, lowercased, punctuation removed."""
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    return text.split()

# Sample text: Genesis 1-2 (creation narrative)
genesis_creation = """
In the beginning God created the heaven and the earth.
And the earth was without form, and void; and darkness was upon the face of the deep.
And the Spirit of God moved upon the face of the waters.
And God said, Let there be light: and there was light.
And God saw the light, that it was good: and God divided the light from the darkness.
And God called the light Day, and the darkness he called Night.
And the evening and the morning were the first day.
And God said, Let there be a firmament in the midst of the waters,
and let it divide the waters from the waters.
And God made the firmament, and divided the waters which were under the firmament
from the waters which were above the firmament: and it was so.
And God called the firmament Heaven. And the evening and the morning were the second day.
And God said, Let the waters under the heaven be gathered together unto one place,
and let the dry land appear: and it was so.
And God called the dry land Earth; and the gathering together of the waters called he Seas:
and God saw that it was good.
And God said, Let the earth bring forth grass, the herb yielding seed,
and the fruit tree yielding fruit after his kind, whose seed is in itself, upon the earth: and it was so.
And the earth brought forth grass, and herb yielding seed after his kind,
and the tree yielding fruit, whose seed was in itself, after his kind: and God saw that it was good.
And the evening and the morning were the third day.
And God said, Let there be lights in the firmament of the heaven to divide the day from the night;
and let them be for signs, and for seasons, and for days, and years:
And let them be for lights in the firmament of the heaven to give light upon the earth: and it was so.
And God made two great lights; the greater light to rule the day, and the lesser light to rule the night:
he made the stars also.
And God set them in the firmament of the heaven to give light upon the earth,
And to rule over the day and over the night, and to divide the light from the darkness:
and God saw that it was good.
And the evening and the morning were the fourth day.
And God said, Let the waters bring forth abundantly the moving creature that hath life,
and fowl that may fly above the earth in the open firmament of heaven.
And God created great whales, and every living creature that moveth,
which the waters brought forth abundantly, after their kind, and every winged fowl after his kind:
and God saw that it was good.
And God blessed them, saying, Be fruitful, and multiply, and fill the waters in the seas,
and let fowl multiply in the earth.
And the evening and the morning were the fifth day.
And God said, Let the earth bring forth the living creature after his kind,
cattle, and creeping thing, and beast of the earth after his kind: and it was so.
And God made the beast of the earth after his kind, and cattle after their kind,
and every thing that creepeth upon the earth after his kind: and God saw that it was good.
And God said, Let us make man in our image, after our likeness:
and let them have dominion over the fish of the sea, and over the fowl of the air,
and over the cattle, and over all the earth, and over every creeping thing that creepeth upon the earth.
So God created man in his own image, in the image of God created he him; male and female created he them.
And God blessed them, and God said unto them, Be fruitful, and multiply, and replenish the earth,
and subdue it: and have dominion over the fish of the sea, and over the fowl of the air,
and over every living thing that moveth upon the earth.
"""

words = get_words(genesis_creation)
print(f"Loaded {len(words)} words")
```

## 2. Defining Motif Buckets

A motif bucket is just a **list of words** that we decide belong together.

Here are some buckets for analyzing creation narratives:

```python
# Define motif buckets
# Each bucket is a list of words that we're grouping together

MOTIF_BUCKETS = {
    "light_and_dark": [
        "light", "lights", "bright", "shine", "sun", "day", "morning",
        "dark", "darkness", "night", "evening", "shadow"
    ],
    
    "water": [
        "water", "waters", "sea", "seas", "deep", "flood", "rain",
        "river", "stream", "wave", "ocean"
    ],
    
    "earth_and_land": [
        "earth", "land", "ground", "dust", "soil", "dry", "clay"
    ],
    
    "living_things": [
        "creature", "creatures", "living", "life", "beast", "beasts",
        "cattle", "fowl", "fish", "bird", "birds", "animal", "animals",
        "whale", "whales", "creeping", "creepeth"
    ],
    
    "plants": [
        "grass", "herb", "tree", "trees", "fruit", "seed", "plant",
        "plants", "leaf", "leaves", "flower", "garden"
    ],
    
    "divine_action": [
        "god", "lord", "said", "made", "created", "called", "saw",
        "blessed", "divided", "set"
    ],
    
    "human": [
        "man", "woman", "male", "female", "image", "likeness",
        "dominion", "subdue", "adam", "eve"
    ],
    
    "time_and_order": [
        "day", "days", "first", "second", "third", "fourth", "fifth",
        "sixth", "seventh", "beginning", "seasons", "years"
    ]
}

print("Defined motif buckets:")
for name, words_list in MOTIF_BUCKETS.items():
    print(f"  {name}: {len(words_list)} words")
```

### Notice

Some words appear in multiple buckets. "day" is in both `light_and_dark` and `time_and_order`.

This is a **choice**. You could decide that "day" only belongs to one bucket, or that it belongs to both.

There's no correct answer—only interpretive consequences.

---

## 3. Counting Motifs

Now we count how many words from each bucket appear in the text.

```python
def count_motif(words, bucket):
    """Count how many words from a bucket appear in the text."""
    count = 0
    matches = []
    for word in words:
        if word in bucket:
            count += 1
            matches.append(word)
    return count, matches

def analyze_motifs(words, buckets):
    """Analyze all motif buckets for a text."""
    results = {}
    for name, bucket in buckets.items():
        count, matches = count_motif(words, bucket)
        results[name] = {
            "count": count,
            "matches": Counter(matches)
        }
    return results

# Analyze Genesis
motif_results = analyze_motifs(words, MOTIF_BUCKETS)

print("Motif counts in Genesis creation narrative:\n")
for name, data in sorted(motif_results.items(), key=lambda x: x[1]["count"], reverse=True):
    print(f"{name}: {data['count']} occurrences")
```

```python
# See which specific words matched
print("Breakdown by motif:\n")
for name, data in motif_results.items():
    if data["count"] > 0:
        print(f"--- {name} ({data['count']} total) ---")
        for word, count in data["matches"].most_common():
            print(f"  {word}: {count}")
        print()
```

### What do you see?

The motif counts give you a **profile** of the text's concerns:
- Which motifs dominate?
- Which are surprisingly absent?
- How does this match your intuition from reading?

## 4. Visualizing the Motif Profile

A simple bar chart makes the pattern clearer.

```python
# Simple text-based bar chart (no matplotlib needed)
def text_bar_chart(data, max_width=40):
    """Print a simple text-based bar chart."""
    max_val = max(data.values()) if data else 1
    for name, value in sorted(data.items(), key=lambda x: x[1], reverse=True):
        bar_length = int((value / max_val) * max_width)
        bar = "█" * bar_length
        print(f"{name:20} {bar} {value}")

# Create summary data
motif_counts = {name: data["count"] for name, data in motif_results.items()}

print("Motif Profile: Genesis Creation\n")
text_bar_chart(motif_counts)
```

## 5. Comparing Texts by Motif

Motif profiles become powerful when you compare texts. Let's add another creation narrative.

```python
# Hesiod's Theogony (cosmogony section)
theogony = """
Verily at the first Chaos came to be, but next wide-bosomed Earth,
the ever-sure foundation of all the deathless ones who hold the peaks of snowy Olympus,
and dim Tartarus in the depth of the wide-pathed Earth,
and Eros, fairest among the deathless gods, who unnerves the limbs
and overcomes the mind and wise counsels of all gods and all men within them.
From Chaos came forth Erebus and black Night;
but of Night were born Aether and Day, whom she conceived and bore from union in love with Erebus.
And Earth first bore starry Heaven, equal to herself,
to cover her on every side, and to be an ever-sure abiding-place for the blessed gods.
And she brought forth long Hills, graceful haunts of the goddess Nymphs who dwell amongst the glens of the hills.
She bore also the fruitless deep with his raging swell, Pontus, without sweet union of love.
But afterwards she lay with Heaven and bore deep-swirling Oceanus,
Coeus and Crius and Hyperion and Iapetus,
Theia and Rhea, Themis and Mnemosyne and gold-crowned Phoebe and lovely Tethys.
After them was born Cronos the wily, youngest and most terrible of her children,
and he hated his lusty sire.
And again, she bore the Cyclopes, overbearing in spirit,
Brontes, and Steropes and stubborn-hearted Arges,
who gave Zeus the thunder and made the thunderbolt:
in all else they were like the gods, but one eye only was set in the midst of their foreheads.
And they were surnamed Cyclopes because one orbed eye was set in their foreheads.
Strength and might and craft were in their works.
"""

theogony_words = get_words(theogony)
theogony_motifs = analyze_motifs(theogony_words, MOTIF_BUCKETS)
theogony_counts = {name: data["count"] for name, data in theogony_motifs.items()}

print("Motif Profile: Hesiod's Theogony\n")
text_bar_chart(theogony_counts)
```

```python
# Side-by-side comparison
print("Comparison: Genesis vs Theogony\n")
print(f"{'Motif':20} {'Genesis':>10} {'Theogony':>10} {'Difference':>12}")
print("-" * 55)

for name in MOTIF_BUCKETS.keys():
    g = motif_counts.get(name, 0)
    t = theogony_counts.get(name, 0)
    diff = g - t
    diff_str = f"+{diff}" if diff > 0 else str(diff)
    print(f"{name:20} {g:>10} {t:>10} {diff_str:>12}")
```

### Interpretation

What does the comparison reveal?

- **Genesis** emphasizes: divine action, light/dark cycles, living creatures, plants
- **Theogony** emphasizes: earth/land, divine genealogy (not captured well by our buckets)

The buckets we chose shape what we can see. Theogony is full of **names** and **genealogies**—but we didn't make a bucket for that.

## 6. Revising Your Buckets

This is the interpretive loop:

1. Define buckets based on what you think matters
2. Run the analysis
3. Notice what's missing or wrong
4. Revise the buckets
5. Repeat

Let's add a bucket that might capture more of Theogony:

```python
# Add new buckets
REVISED_BUCKETS = MOTIF_BUCKETS.copy()

REVISED_BUCKETS["birth_and_generation"] = [
    "born", "bore", "brought", "forth", "conceived", "children",
    "son", "daughter", "father", "mother", "sire", "offspring"
]

REVISED_BUCKETS["cosmic_structure"] = [
    "heaven", "heavens", "sky", "firmament", "chaos", "tartarus",
    "olympus", "abyss", "void", "deep"
]

# Re-analyze
genesis_revised = analyze_motifs(words, REVISED_BUCKETS)
theogony_revised = analyze_motifs(theogony_words, REVISED_BUCKETS)

print("Revised comparison:\n")
print(f"{'Motif':20} {'Genesis':>10} {'Theogony':>10}")
print("-" * 42)

for name in REVISED_BUCKETS.keys():
    g = genesis_revised[name]["count"]
    t = theogony_revised[name]["count"]
    print(f"{name:20} {g:>10} {t:>10}")
```

### Reflection

Adding `birth_and_generation` likely shows Theogony's emphasis on divine genealogy.

But notice: **the buckets you choose determine what you can see.**

This is true of all analytical frameworks, not just word buckets.

## 7. Motifs Over Time (Position in Text)

Where do motifs appear? Beginning, middle, end?

This can reveal narrative structure without reading for plot.

```python
def motif_positions(words, bucket):
    """Find where in the text (as percentage) motif words appear."""
    total = len(words)
    positions = []
    for i, word in enumerate(words):
        if word in bucket:
            position = i / total  # 0.0 = beginning, 1.0 = end
            positions.append(position)
    return positions

# Check where light/dark motif appears in Genesis
light_positions = motif_positions(words, MOTIF_BUCKETS["light_and_dark"])
living_positions = motif_positions(words, MOTIF_BUCKETS["living_things"])
human_positions = motif_positions(words, MOTIF_BUCKETS["human"])

def describe_distribution(positions, name):
    if not positions:
        print(f"{name}: no occurrences")
        return
    avg = sum(positions) / len(positions)
    early = sum(1 for p in positions if p < 0.33)
    middle = sum(1 for p in positions if 0.33 <= p < 0.66)
    late = sum(1 for p in positions if p >= 0.66)
    print(f"{name}:")
    print(f"  Average position: {avg:.1%} through text")
    print(f"  Early (0-33%): {early}, Middle (33-66%): {middle}, Late (66-100%): {late}")

describe_distribution(light_positions, "light_and_dark")
print()
describe_distribution(living_positions, "living_things")
print()
describe_distribution(human_positions, "human")
```

### What does this show?

In Genesis 1:
- **Light/dark** appears early (Days 1, 4)
- **Living things** appear in the middle-to-late section (Days 5-6)
- **Human** appears at the end (Day 6)

This matches the narrative structure—but we found it by counting, not reading.

## Summary: Buckets as Theory

| Concept | What It Means |
|---------|---------------|
| **Motif bucket** | A list of words you decide belong together |
| **Motif count** | How many bucket words appear in a text |
| **Motif profile** | The pattern of counts across all buckets |
| **Bucket revision** | Changing your theory based on what you see |

**Key insight**: The buckets are not neutral. They encode your assumptions about what matters.
