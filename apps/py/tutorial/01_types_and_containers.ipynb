# Types & Containers: What Kind of Thing Is This?

Before we ask a model to describe a poem, we need to understand **what kinds of descriptions exist**.

This notebook introduces:
- **Strings**: raw text
- **Numbers**: counts and measurements
- **Booleans**: yes/no answers
- **Lists**: collections of things
- **Dictionaries**: named fields
- **Enums**: controlled vocabularies

---

## Why bother with types?

Imagine asking ten people: *"Is this poem dark?"*

You might get:
- "Kind of"
- "Yes, very"
- "It's more melancholy than dark"
- "7/10 dark"
- "Depends what you mean by dark"

These answers are all *valid interpretations*, but they're **incomparable**. You can't average them, sort them, or combine them.

**Structure = agreements about how to store answers so they can be reused.**

---

## 1. Strings: Raw Text

A string is just text. It's the most flexible container—and therefore the least structured.

```python
poem_text = """The fog comes
on little cat feet.

It sits looking
over harbor and city
on silent haunches
and then moves on."""

print(type(poem_text))
print(poem_text)
```

A string is good for:
- Storing the original poem
- Free-form notes
- Anything where you don't know the structure yet

A string is bad for:
- Counting things
- Comparing across poems
- Anything that needs to be "the same shape" every time

---

## 2. Numbers: Counts and Measurements

Numbers let you measure, compare, and compute.

```python
line_count = 7
word_count = 21
avg_words_per_line = word_count / line_count

print(f"Lines: {line_count}")
print(f"Words: {word_count}")
print(f"Average words per line: {avg_words_per_line:.2f}")
```

Two kinds of numbers:
- **int** (integer): whole numbers like `7`, `14`, `0`
- **float**: decimals like `3.0`, `2.5`, `0.75`

Numbers are great for:
- Line counts, word counts, syllable counts
- Ratios and proportions
- Anything you want to average or compare numerically

---

## 3. Booleans: Yes or No

A boolean is the simplest structured answer: `True` or `False`.

```python
has_rhyme = False
has_metaphor = True
is_sonnet = False

print(f"Has rhyme: {has_rhyme}")
print(f"Has metaphor: {has_metaphor}")
print(f"Is sonnet: {is_sonnet}")
```

Booleans force a decision. No "kind of" allowed.

This is a feature, not a bug: **constraints create comparability**.

---

## 4. Lists: Multiple Things of the Same Kind

A list holds multiple values. Order matters (first, second, third...).

```python
imagery = ["fog", "cat", "harbor", "city", "haunches"]

print(type(imagery))
print(f"Number of images: {len(imagery)}")
print(f"First image: {imagery[0]}")
print(f"All images: {imagery}")
```

### String ↔ List: Two Views of the Same Poem

Watch how we can transform between containers:

```python
# String → List (split by lines)
lines = poem_text.splitlines()
print(f"Type: {type(lines)}")
print(f"Number of lines: {len(lines)}")
print(lines)
```

```python
# List → String (join back together)
rejoined = "\n".join(lines)
print(f"Type: {type(rejoined)}")
print(rejoined)
```

**Key insight**: The container you choose affects what operations are easy.

- **String-friendly tasks**: find repeated words, count punctuation
- **List-friendly tasks**: count lines, find average line length, access specific lines

## 5. Dictionaries: Named Fields

A dictionary (dict) stores **key-value pairs**. Each piece of data has a name.

```python
poem_profile = {
    "title": "Fog",
    "author": "Carl Sandburg",
    "line_count": 7,
    "has_rhyme": False,
    "imagery": ["fog", "cat", "harbor", "city"],
    "tone": "contemplative"
}

print(type(poem_profile))
print(poem_profile)
```

```python
# Access individual fields
print(f"Title: {poem_profile['title']}")
print(f"Line count: {poem_profile['line_count']}")
print(f"Imagery: {poem_profile['imagery']}")
```

Notice: a dictionary can contain **different types** of values:
- `"title"` → string
- `"line_count"` → int
- `"has_rhyme"` → boolean
- `"imagery"` → list

This is what makes dictionaries powerful for **structured descriptions**.

## 6. Enums: Controlled Vocabulary

An enum is a **restricted set of allowed values**.

Why? Because free-text answers are chaos:

```python
---

## Next: Repetition and Formula

Now that you know the building blocks, we'll use them to count words and find patterns in texts.

→ Continue to `02_repetition_and_formula.ipynb`
```

```python
# With a controlled vocabulary:
ALLOWED_TONES = [
    "awe",
    "grief",
    "irony",
    "rage",
    "tenderness",
    "contemplation",
    "longing",
    "defiance",
    "wonder",
    "despair"
]

def validate_tone(tone):
    if tone in ALLOWED_TONES:
        return True
    else:
        print(f"'{tone}' is not allowed. Choose from: {ALLOWED_TONES}")
        return False

validate_tone("contemplation")  # ✓
validate_tone("vibes")          # ✗
```

**Enums create comparability.** If everyone uses the same vocabulary, you can:
- Count how many poems have "grief" as their tone
- Find all poems with "irony"
- Compare your choices to someone else's

The trade-off: you lose nuance. "Sad-ish" might be more accurate, but it's useless for comparison.

## Summary: The Type Hierarchy

| Type | What it holds | Example |
|------|--------------|--------|
| `str` | Raw text | `"The fog comes"` |
| `int` | Whole numbers | `7` |
| `float` | Decimals | `3.14` |
| `bool` | Yes/No | `True` |
| `list` | Multiple values | `["fog", "cat", "harbor"]` |
| `dict` | Named fields | `{"title": "Fog", "lines": 7}` |
| enum | Controlled choices | One of `["awe", "grief", "irony", ...]` |

---

## The Category Theory Whisper 🔮

*You don't need to know this, but if you're curious...*

What we've been doing is thinking about **objects** (kinds of data) and **morphisms** (transformations between them).

- `poem_text.splitlines()` is a morphism: `str → list[str]`
- `"\n".join(lines)` is a morphism: `list[str] → str`
- Building a profile is a morphism: `str → dict`

When the output type of one function matches the input type of another, you can **compose** them into a pipeline.

That's the secret: **types are agreements that make composition possible.**
