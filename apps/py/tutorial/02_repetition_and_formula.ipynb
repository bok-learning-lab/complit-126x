# Repetition & Formula: What a Text Says Over and Over

Literature repeats itself. Words, phrases, images, syntactic patterns—repetition is how texts create rhythm, emphasis, and meaning.

This notebook introduces **word frequency** as a way to see what a text insists on.

---

## The Question

If you had to describe a text using only its **most repeated words**, what would you learn?

What would you miss?

---

## 1. Setup: Loading a Text

We'll work with a sample text embedded here. Later, you can load texts from the course readings folder.

```python
from collections import Counter
import re

# Sample text: Genesis 1 (KJV)
# Later this can be loaded from: ../../_content/readings/genesis.md

genesis_1 = """
In the beginning God created the heaven and the earth.
And the earth was without form, and void; and darkness was upon the face of the deep.
And the Spirit of God moved upon the face of the waters.
And God said, Let there be light: and there was light.
And God saw the light, that it was good: and God divided the light from the darkness.
And God called the light Day, and the darkness he called Night.
And the evening and the morning were the first day.
And God said, Let there be a firmament in the midst of the waters,
and let it divide the waters from the waters.
And God made the firmament, and divided the waters which were under the firmament
from the waters which were above the firmament: and it was so.
And God called the firmament Heaven. And the evening and the morning were the second day.
And God said, Let the waters under the heaven be gathered together unto one place,
and let the dry land appear: and it was so.
And God called the dry land Earth; and the gathering together of the waters called he Seas:
and God saw that it was good.
And God said, Let the earth bring forth grass, the herb yielding seed,
and the fruit tree yielding fruit after his kind, whose seed is in itself, upon the earth: and it was so.
And the earth brought forth grass, and herb yielding seed after his kind,
and the tree yielding fruit, whose seed was in itself, after his kind: and God saw that it was good.
And the evening and the morning were the third day.
"""

print(f"Loaded {len(genesis_1)} characters")
print(genesis_1[:200] + "...")
```

---

## 2. Counting Words

The simplest distant reading move: count how often each word appears.

```python
def get_words(text):
    """Extract words from text, lowercased, punctuation removed."""
    # Remove punctuation and lowercase
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    # Split into words
    words = text.split()
    return words

words = get_words(genesis_1)
print(f"Total words: {len(words)}")
print(f"First 20 words: {words[:20]}")
```

```python
# Count word frequencies
word_counts = Counter(words)

print("Top 20 most frequent words:")
for word, count in word_counts.most_common(20):
    print(f"  {word}: {count}")
```

### What do you notice?

The most frequent words are probably **function words**: "the", "and", "was", "of"...

These words are grammatically necessary but don't tell us much about *this* text specifically.

They're sometimes called **stopwords**.

---

## 3. Filtering Stopwords

Let's remove common English words to see what's left.

```python
# A simple stopword list (you can modify this!)
STOPWORDS = {
    'the', 'and', 'a', 'an', 'of', 'to', 'in', 'is', 'it', 'that', 'was', 'were',
    'be', 'been', 'being', 'have', 'has', 'had', 'do', 'does', 'did',
    'for', 'on', 'with', 'as', 'at', 'by', 'from', 'or', 'but', 'not',
    'this', 'which', 'there', 'their', 'they', 'he', 'she', 'his', 'her',
    'i', 'you', 'we', 'my', 'your', 'our', 'its', 'who', 'what', 'when',
    'where', 'why', 'how', 'all', 'each', 'every', 'both', 'few', 'more',
    'most', 'other', 'some', 'such', 'no', 'nor', 'only', 'own', 'same',
    'so', 'than', 'too', 'very', 'just', 'can', 'will', 'should', 'now',
    'unto', 'upon', 'let', 'whose', 'after', 'itself'
}

def filter_stopwords(words):
    return [w for w in words if w not in STOPWORDS]

content_words = filter_stopwords(words)
print(f"Words after filtering: {len(content_words)} (was {len(words)})")
```

```python
# Count again without stopwords
content_counts = Counter(content_words)

print("Top 20 content words:")
for word, count in content_counts.most_common(20):
    print(f"  {word}: {count}")
```

### Now what do you see?

These words are more *about* something. They're the vocabulary of this particular text.

In Genesis 1, you might see: **god**, **earth**, **waters**, **light**, **day**, **evening**, **morning**, **firmament**...

This is already a kind of summary—a lossy compression of the text into its most insisted-upon terms.

---

## 4. The Mid-Frequency Zone

The most frequent words are often obvious. The least frequent words are often noise.

The **mid-frequency** words—appearing 2-5 times—are often the most interesting. They're distinctive but not dominant.

```python
# Find mid-frequency words
mid_freq = [(word, count) for word, count in content_counts.items() if 2 <= count <= 5]
mid_freq.sort(key=lambda x: x[1], reverse=True)

print(f"Mid-frequency words (appearing 2-5 times): {len(mid_freq)} words")
print()
for word, count in mid_freq:
    print(f"  {word}: {count}")
```

### Reflection

Mid-frequency words often reveal:
- **Motifs** that recur but don't dominate
- **Specific vocabulary** that distinguishes this text from others
- **Thematic concerns** that aren't immediately obvious

In Genesis 1, mid-frequency words might include: **seed**, **kind**, **fruit**, **herb**, **dry**, **divided**...

---

## 5. Formula: Repeated Phrases

Words repeat, but so do **phrases**. Formulaic language is a signature of oral tradition, liturgy, and certain literary styles.

Let's look for repeated two-word and three-word sequences.

```python
def get_ngrams(words, n):
    """Get all n-word sequences from a list of words."""
    ngrams = []
    for i in range(len(words) - n + 1):
        ngram = ' '.join(words[i:i+n])
        ngrams.append(ngram)
    return ngrams

# Get bigrams (2-word phrases)
bigrams = get_ngrams(words, 2)
bigram_counts = Counter(bigrams)

print("Most repeated 2-word phrases:")
for phrase, count in bigram_counts.most_common(15):
    if count > 1:  # Only show repeated phrases
        print(f"  '{phrase}': {count}")
```

```python
# Get trigrams (3-word phrases)
trigrams = get_ngrams(words, 3)
trigram_counts = Counter(trigrams)

print("Most repeated 3-word phrases:")
for phrase, count in trigram_counts.most_common(15):
    if count > 1:
        print(f"  '{phrase}': {count}")
```

### What formulas do you find?

In Genesis 1, you might see:
- "and god said"
- "it was so"
- "and god saw"
- "the evening and the morning"
- "after his kind"

These are the **structural formulas** of the text—the repeated scaffolding that organizes the narrative.

Oral traditions rely heavily on formula. So do legal codes, liturgies, and certain modernist experiments.

---

## 6. Hapax Legomena: Words That Appear Only Once

The opposite of repetition: words that appear **exactly once**.

In biblical studies, these are called *hapax legomena* (Greek: "said once"). They're often the hardest words to translate because there's no other context.

```python
# Find words that appear only once
hapax = [word for word, count in content_counts.items() if count == 1]

print(f"Words appearing only once: {len(hapax)}")
print()
print("Examples:")
for word in sorted(hapax)[:20]:
    print(f"  {word}")
```

### Reflection

Hapax words are:
- **Unique** to a specific moment in the text
- Often **specific** (proper nouns, technical terms)
- Sometimes **errors** or **variants**

A text with many hapax words is **lexically diverse**. A text with few is **formulaic**.

## 7. Comparing Two Texts

Frequency becomes more interesting when you compare. What does one text repeat that another doesn't?

```python
# A second sample text: Hesiod's Theogony (opening)
theogony_opening = """
From the Heliconian Muses let us begin to sing,
who hold the great and holy mount of Helicon,
and dance on soft feet about the deep-blue spring
and the altar of the almighty son of Cronos,
and, when they have washed their tender bodies in Permessus
or in the Horse's Spring or Olmeius,
make their fair, lovely dances upon highest Helicon
and move with vigorous feet.
Thence they arise and go abroad by night,
veiled in thick mist, and utter their song with lovely voice,
praising Zeus the aegis-holder, and queenly Hera
of Argos who walks on golden sandals,
and the daughter of Zeus the aegis-holder, bright-eyed Athena,
and Phoebus Apollo, and Artemis who delights in arrows,
and Poseidon the earth-holder who shakes the earth,
and reverend Themis, and quick-glancing Aphrodite,
and Hebe with the crown of gold, and fair Dione,
Leto, Iapetus, and Cronos the crafty counsellor,
Eos and great Helius and bright Selene,
Earth too, and great Oceanus, and dark Night,
and the holy race of all the other deathless ones that are for ever.
"""

print(f"Loaded Theogony excerpt: {len(theogony_opening)} characters")
```

```python
# Compare word frequencies
theogony_words = get_words(theogony_opening)
theogony_content = filter_stopwords(theogony_words)
theogony_counts = Counter(theogony_content)

print("Genesis 1 - Top 10 content words:")
for word, count in content_counts.most_common(10):
    print(f"  {word}: {count}")

print("\nTheogony - Top 10 content words:")
for word, count in theogony_counts.most_common(10):
    print(f"  {word}: {count}")
```

```python
# Words unique to each text
genesis_vocab = set(content_counts.keys())
theogony_vocab = set(theogony_counts.keys())

only_genesis = genesis_vocab - theogony_vocab
only_theogony = theogony_vocab - genesis_vocab
shared = genesis_vocab & theogony_vocab

print(f"Words only in Genesis: {len(only_genesis)}")
print(f"  Examples: {list(only_genesis)[:10]}")

print(f"\nWords only in Theogony: {len(only_theogony)}")
print(f"  Examples: {list(only_theogony)[:10]}")

print(f"\nShared words: {len(shared)}")
print(f"  {list(shared)}")
```

### Reflection

Both texts are creation narratives. Both are ancient. Both are foundational.

But their vocabularies diverge:
- Genesis repeats **god**, **earth**, **waters**, **light**, **day**
- Theogony repeats **zeus**, **helicon**, **muses**, names of gods

What does this tell you about how each text imagines creation?

## Summary: What Frequency Reveals

| Frequency Zone | What It Shows |
|----------------|---------------|
| **Very high** (stopwords) | Grammar, syntax, language itself |
| **High** (content words) | What the text insists on, its obsessions |
| **Mid** (2-5 occurrences) | Motifs, distinctive vocabulary, thematic texture |
| **Low** (hapax) | Specificity, uniqueness, possible errors |
| **Repeated phrases** | Formula, structure, oral/liturgical patterns |
