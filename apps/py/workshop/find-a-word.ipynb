````python
import re

def count_word_occurrences(text, word):
    """
    Counts the number of times a word appears in a given text.
    The search is case-insensitive and considers whole words only.
    """
    # Use re.findall for case-insensitive whole word matching
    matches = re.findall(r'\b' + re.escape(word) + r'\b', text, re.IGNORECASE)
    return len(matches)

# Example usage:
search_word = "God"
occurrences = count_word_occurrences(genesis_1_text, search_word)
print(f"The word '{search_word}' appears {occurrences} times in Genesis 1.")

search_word_2 = "earth"
occurrences_2 = count_word_occurrences(genesis_1_text, search_word_2)
print(f"The word '{search_word_2}' appears {occurrences_2} times in Genesis 1.")
````

### Understanding Python Functions for Beginners

In Python, a **function** is like a mini-program or a recipe that you can define once and then use multiple times. It's a block of organized, reusable code that performs a single, related action.

Let's break down the `count_word_occurrences` function we just used:

```python
def count_word_occurrences(text, phrase):
    # This is the function's body
    matches = re.findall(r'(?i)' + re.escape(phrase), text)
    return len(matches)
```

1.  **`def` keyword**: This keyword tells Python that you are about to *define* a new function.

2.  **`count_word_occurrences`**: This is the **name** of the function. You choose a name that describes what the function does (e.g., `count_word_occurrences` counts word occurrences, `print_hello` would print "hello").

3.  **`(text, phrase)`**: These are the **arguments** (also called parameters). Arguments are pieces of information that you *pass into* the function so it can use them to do its job. Think of them as ingredients for your recipe:
    *   `text`: This argument is where the function expects the full text (like Genesis 1) to be provided.
    *   `phrase`: This argument is where the function expects the word or phrase you want to count.

4.  **`:` (colon)**: The colon marks the end of the `def` line and indicates that the indented code block that follows is the *body* of the function.

5.  **Indentation**: The lines of code that belong to the function must be indented (usually with 4 spaces). Python uses indentation to define code blocks.

6.  **`# This is the function's body`**: This is a **comment**. Comments are lines of text in code that Python ignores. They are there to help humans understand what the code is doing.

7.  **`return len(matches)`**: This is the **return statement**. After the function has finished its work, it can `return` a value back to the part of the code that called it. In our case, `len(matches)` calculates the number of matches found, and this numerical count is the value the function *gives back*.

### How to use (Call) a Function

To make the function run, you **call** it by its name, providing the necessary arguments inside parentheses:

```python
search_word = "God"
occurrences = count_word_occurrences(genesis_1_text, search_word)
# When this line runs:
#   - The text in `genesis_1_text` is passed as `text` to the function.
#   - The value of `search_word` ("God") is passed as `phrase` to the function.
#   - The function calculates the count and 'returns' it.
#   - The returned value is then stored in the `occurrences` variable.
print(f"The word '{search_word}' appears {occurrences} times in Genesis 1.")
```

Functions help keep your code organized, readable, and prevent you from writing the same code over and over again!
