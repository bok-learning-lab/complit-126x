## The Key Concepts

Here's what you just learned:

| Concept | What it means |
|---------|---------------|
| **Function** | A reusable block of code with a name. You "call" it to run it. |
| **Arguments** | The inputs you pass to a function (in parentheses). |
| **Return value** | What the function gives back when it's done. |

The pattern looks like this:

```python
def function_name(argument1, argument2):
    # do something
    return result
```

And you use it like this:

```python
result = function_name(value1, value2)
```
