Let's start with three texts to work with:

````python
# Three passages from the King James Bible

genesis = """In the beginning God created the heaven and the earth. And the earth was without form, and void; and darkness was upon the face of the deep. And the Spirit of God moved upon the face of the waters. And God said, Let there be light: and there was light. And God saw the light, that it was good: and God divided the light from the darkness."""

exodus = """And God spake all these words, saying, I am the LORD thy God, which have brought thee out of the land of Egypt, out of the house of bondage. Thou shalt have no other gods before me. Thou shalt not make unto thee any graven image, or any likeness of any thing that is in heaven above, or that is in the earth beneath, or that is in the water under the earth."""

psalms = """The LORD is my shepherd; I shall not want. He maketh me to lie down in green pastures: he leadeth me beside the still waters. He restoreth my soul: he leadeth me in the paths of righteousness for his name's sake. Yea, though I walk through the valley of the shadow of death, I will fear no evil: for thou art with me; thy rod and thy staff they comfort me."""
````

And here's the function we made in the last notebook:

````python
def count_word(text, word):
    """Count how many times a word appears in a text (case-insensitive)."""
    text_lower = text.lower()
    word_lower = word.lower()
    words = text_lower.split()
    count = 0
    for w in words:
        w_clean = w.strip('.,;:!?"\'')
        if w_clean == word_lower:
            count += 1
    return count
````

## The Key Concepts

| Concept | What it means |
|---------|---------------|
| **List** | A collection of items in order: `[item1, item2, item3]` |
| **For loop** | "Do this for each item in the list" |
| **Iteration** | Each pass through the loop |

The pattern:

```python
for item in my_list:
    # do something with item
```
