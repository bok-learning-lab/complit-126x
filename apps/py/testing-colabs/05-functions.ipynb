# 05 - Functions

A **function** is reusable code with a name.

```python
def my_function(input):
    # do something
    return result
```

Define it once, use it many times.

````
def describe(poem_title, bird):
    return f"{poem_title} features a {bird}"

print(describe("The Raven", "raven"))
print(describe("Ode to a Nightingale", "nightingale"))
````

Functions can do more complex things:

````
def analyze(title, text):
    word_count = len(text.split())
    return f"{title}: {word_count} words"

result = analyze("The Raven", "Once upon a midnight dreary, while I pondered, weak and weary")
print(result)
````

Use functions in loops:

````
poems = [
    {"title": "The Raven", "bird": "raven"},
    {"title": "Ode to a Nightingale", "bird": "nightingale"},
    {"title": "The Windhover", "bird": "falcon"},
]

for p in poems:
    result = describe(p["title"], p["bird"])
    print(result)
````
