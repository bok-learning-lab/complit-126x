# 03 - Loops

A **loop** does something to each item in a list.

```python
for item in my_list:
    do_something(item)
```

The indented code runs once for each item.

````
poems = ["The Raven", "Ode to a Nightingale", "Hope is the thing with feathers"]

for poem in poems:
    print(poem)
````

You can do more inside the loop:

````
for poem in poems:
    print(f"Analyzing: {poem}")
    print(f"  Title has {len(poem)} characters")
    print()
````

---

## Why this matters

Instead of writing the same code 3 times (or 100 times), you write it once and loop.
