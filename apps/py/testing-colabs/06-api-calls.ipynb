# 06 - API Calls

An **API call** sends a request to an AI and gets a response.

It's like texting ChatGPT from your code.

```
from openai import OpenAI
client = OpenAI()
```

```
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "What bird appears in Poe's famous poem? One sentence."}]
)

print(response.choices[0].message.content)
```

Wrap it in a function:

```
def ask(question):
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": question}]
    )
    return response.choices[0].message.content

print(ask("What bird is in Keats' Ode to a Nightingale? One sentence."))
```

Use it in a loop:

```
poems = ["The Raven", "Ode to a Nightingale", "The Windhover"]

for poem in poems:
    answer = ask(f"What bird appears in the poem '{poem}'? One sentence.")
    print(f"{poem}: {answer}")
    print()
```
