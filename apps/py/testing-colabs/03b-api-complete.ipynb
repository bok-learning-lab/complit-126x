# 03 - API Calls

## Why Code Instead of ChatGPT? Reason #2: CHAINING

In ChatGPT, each conversation is isolated. If you want to:
1. Summarize a passage
2. Then identify the key themes from that summary
3. Then ask a follow-up question about those themes

...you'd have to copy-paste between steps manually.

With code, you can **chain** these steps together into an automated workflow. The output of step 1 becomes the input to step 2, and so on.

---

## Setup

First, let's make sure the API is ready. Your `OPENAI_API_KEY` should be set in Codespaces Secrets.

````python
import os

if not os.environ.get("OPENAI_API_KEY"):
    raise EnvironmentError(
        "OPENAI_API_KEY not found!\n\n"
        "Add it to Colab Secrets (key icon in left sidebar)"
    )

print("✓ API key found")

from openai import OpenAI
client = OpenAI()
print("✓ OpenAI client ready")
````

And here's a passage to work with:

````python
passage = """In the beginning God created the heaven and the earth. And the earth was without form, and void; and darkness was upon the face of the deep. And the Spirit of God moved upon the face of the waters. And God said, Let there be light: and there was light. And God saw the light, that it was good: and God divided the light from the darkness. And God called the light Day, and the darkness he called Night. And the evening and the morning were the first day."""

print(passage)
````

---

## Step 1: Vibecode It

Go to ChatGPT/Claude/Gemini and type:

> **Write Python code that uses the OpenAI API to send a passage of text and ask for a one-sentence summary. Use the `gpt-4o-mini` model. I already have `client = OpenAI()` set up.**

Paste the code below:

````python
# Here's what the AI might give you:

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": f"Summarize this passage in one sentence:\n\n{passage}"
        }
    ]
)

summary = response.choices[0].message.content
print(summary)
````

---

## Step 2: Unpack It

Ask the AI:

> **Explain this code to me. What is a 'client'? What are 'messages'? What does the response object contain?**

### Understanding the Code

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}]
)
```

- **`client`** — Your connection to OpenAI's servers. It handles authentication and communication.

- **`chat.completions.create()`** — Asks the AI to complete a conversation.

- **`model`** — Which AI model to use. `gpt-4o-mini` is fast and cheap.

- **`messages`** — A list of conversation messages. Each message has:
  - `role`: who's speaking (`"user"` = you, `"assistant"` = AI)
  - `content`: what they said

- **`response.choices[0].message.content`** — Dig into the response to get the actual text:
  - `choices` = list of possible responses (usually just 1)
  - `[0]` = first choice
  - `.message.content` = the text

---

## The Key Concepts

| Concept | What it means |
|---------|---------------|
| **Client** | Your connection to the OpenAI API |
| **Messages** | The conversation: a list of `{role, content}` pairs |
| **Role: user** | A message from you (the human) |
| **Role: assistant** | A message from the AI |
| **Response** | What comes back from the API |

The pattern:

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": your_prompt}]
)
answer = response.choices[0].message.content
```

---

## Step 3: Make It Reusable

Ask the AI:

> **Wrap this API call in a function called `ask(prompt)` that takes a prompt string and returns the response text.**

````python
def ask(prompt):
    """Send a prompt to the AI and return the response text."""
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}]
    )
    return response.choices[0].message.content
````

Test it:

````python
# Test: ask a simple question
ask("What is the capital of France? Answer in one word.")
````

---

## Step 4: Chain It Together

Now for the powerful part. Let's chain multiple steps:

1. Get a summary of the passage
2. Use that summary to identify the main theme
3. Ask a follow-up question about that theme

Ask the AI:

> **Using my `ask()` function, write code that: (1) summarizes the passage in `passage`, (2) takes that summary and asks what the main theme is, (3) asks a follow-up question about how that theme relates to creation myths in other cultures. Print the result of each step.**

````python
# Step 1: Summarize
summary = ask(f"Summarize this passage in one sentence:\n\n{passage}")
print("SUMMARY:")
print(summary)
print()

# Step 2: Identify theme (using the summary from step 1)
theme = ask(f"Based on this summary, what is the main theme in 2-3 words?\n\n{summary}")
print("THEME:")
print(theme)
print()

# Step 3: Follow-up question (using the theme from step 2)
comparison = ask(f"How does the theme of '{theme}' in Genesis compare to creation myths in other cultures? Give 2-3 examples.")
print("COMPARISON:")
print(comparison)
````

---

## The Power of Chaining

Notice what happened:
- Step 1's output (`summary`) → became Step 2's input
- Step 2's output (`theme`) → became Step 3's input

In ChatGPT, you'd have to copy-paste between each step. With code, it flows automatically.

Now imagine chaining 10 steps, or running this chain on 100 different passages. That's the power of combining **loops** (from notebook 02) with **chaining**.

---

## Step 5: Combine Loops + API

Let's bring it together. Here are three passages:

````python
passages = [
    ("Genesis", """In the beginning God created the heaven and the earth. And the earth was without form, and void; and darkness was upon the face of the deep."""),
    ("Hesiod", """First of all, the Void came into being, next broad-bosomed Earth, the solid and eternal home of all, and Eros, the most beautiful of the immortal gods."""),
    ("Ovid", """Before the sea and lands began to be, before the sky had mantled everything, Nature displayed a single face, which they called Chaos: a raw and undivided mass.""")
]
````

Ask the AI:

> **Loop through the `passages` list and for each one, use the `ask()` function to get a one-sentence summary. Print the name and summary for each.**

````python
# Loop through passages and summarize each:

for name, text in passages:
    summary = ask(f"Summarize this passage in one sentence:\n\n{text}")
    print(f"{name}:")
    print(f"  {summary}")
    print()
````
