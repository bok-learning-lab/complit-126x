# 02 - Lists

A **list** holds multiple items. Use square brackets `[]`.

```python
my_list = ["item1", "item2", "item3"]
```

````
poems = ["The Raven", "Ode to a Nightingale", "Hope is the thing with feathers"]
print(poems)
````

Get one item by its position (starts at 0):

````
print(poems[0])  # First item
print(poems[1])  # Second item
print(poems[2])  # Third item
````

Count how many items:

````
print(len(poems))
````
