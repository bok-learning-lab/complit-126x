<cell_type>markdown</cell_type># 08 - Files

Read and write files to your output folder.

We use `Path` to work with files.

## Writing a file

```
# Write a text file
poem_file = OUTPUT_DIR / "raven-excerpt.txt"
poem_file.write_text("Once upon a midnight dreary, while I pondered, weak and weary")

print(f"Wrote to: {poem_file}")
```

## Reading a file

```
content = poem_file.read_text()
print(content)
```

## Reading all files in a folder

```
# List all .txt files
for filepath in OUTPUT_DIR.glob("*.txt"):
    print(f"Found: {filepath.name}")
    print(f"  Content: {filepath.read_text()[:50]}...")
```

<cell_type>markdown</cell_type>## Writing Markdown

You can write `.md` files for formatted text.

```
md_content = """# Birds in Poetry

## The Raven
*Edgar Allan Poe*

> Once upon a midnight dreary, while I pondered, weak and weary

**Bird:** Raven  
**Symbolism:** Death, loss, the supernatural
"""

md_file = OUTPUT_DIR / "analysis.md"
md_file.write_text(md_content)

print(f"Wrote markdown to: {md_file}")
```
