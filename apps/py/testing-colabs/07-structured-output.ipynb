# 07 - Structured Output

Normal API calls return text. **Structured output** returns organized data.

You define a **schema** — a template for what you want back.

```
from openai import OpenAI
from pydantic import BaseModel

client = OpenAI()
```

Define what you want:

```
class BirdPoem(BaseModel):
    poem_title: str
    poet: str
    bird: str
    symbolism: str
```

Ask the AI to fill it in:

```
response = client.beta.chat.completions.parse(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "Analyze the bird in Poe's 'The Raven'"}],
    response_format=BirdPoem
)

result = response.choices[0].message.parsed
print(result)
```

Access each field:

```
print(f"Poem: {result.poem_title}")
print(f"Poet: {result.poet}")
print(f"Bird: {result.bird}")
print(f"Symbolism: {result.symbolism}")
```
