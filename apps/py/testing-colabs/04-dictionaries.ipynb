# 04 - Dictionaries

A **dictionary** stores key-value pairs. Use curly braces `{}`.

```python
my_dict = {"key": "value", "another_key": "another_value"}
```

Think of it like a labeled container.

````
poem = {
    "title": "The Raven",
    "author": "Edgar Allan Poe",
    "year": 1845,
    "bird": "raven"
}

print(poem)
````

Get a value by its key:

````
print(poem["title"])
print(poem["author"])
print(poem["bird"])
````

A list of dictionaries — this is how we'll store multiple poems:

````
poems = [
    {"title": "The Raven", "author": "Poe", "bird": "raven"},
    {"title": "Ode to a Nightingale", "author": "Keats", "bird": "nightingale"},
    {"title": "Hope is the thing with feathers", "author": "Dickinson", "bird": "unnamed"},
]

for p in poems:
    print(f"{p['title']} by {p['author']} — features a {p['bird']}")
````
